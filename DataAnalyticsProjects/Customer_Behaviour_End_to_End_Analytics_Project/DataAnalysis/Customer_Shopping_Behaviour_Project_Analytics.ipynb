Importing Libraries to perform analysis


````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
````

Loading the Dataset

````python
cust_data=pd.read_csv('/content/customer_shopping_behavior.csv')
cust_data.head(10)
````

````python
#Description of dataset
cust_data.describe(include='all')
````

````python
#Rows and columns of data
cust_data.shape
````

````python
#Extracting the Null Values
cust_data.isnull().sum()
````

````python
#We are replacing the null values of Reviews Ratings columns with Imputing Median Values based on Category column
cust_data['Review Rating']=cust_data.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
````

````python
cust_data.isnull().sum()
````

````python
#Converting the columns into Snake Casing for better analysis in the further process.
cust_data.columns=cust_data.columns.str.lower()
cust_data.columns=cust_data.columns.str.replace(' ','_')
cust_data.columns
cust_data.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
````

**Feature Engineering**

````python
#Create a columns age group based on Ages
labels=['Young Adults','Adult','Middle-Aged','Senior Citizens']
cust_data['age_group']=pd.qcut(cust_data['age'],4,labels=labels)
````

````python
cust_data[['age','age_group']].head(20)
````

````python
#Create a column purchase_frequency_days
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}

cust_data['purchase_frequency_days']=cust_data['frequency_of_purchases'].map(frequency_mapping)
````

````python
cust_data[['frequency_of_purchases','purchase_frequency_days']].head(20)
````

````python
#Performing operation on Discount Applied and promo Code Used
cust_data[['discount_applied','promo_code_used']].head(20)
````

````python
(cust_data['discount_applied']==cust_data['promo_code_used']).all()
````

````python
cust_data=cust_data.drop('promo_code_used',axis=1)
cust_data
````
